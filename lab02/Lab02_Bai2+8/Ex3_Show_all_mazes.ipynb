```python
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
import os
import maze_helper as mh
import matplotlib.pyplot as plt
```

```python
import pandas as pd

def list_mazes(directory):
    return sorted([f for f in os.listdir(directory) if f.endswith('.txt')])

mazes = list_mazes('.')
summary=[]
for maze_file in mazes:
    with open(maze_file, 'r') as f:
        maze_str=f.read()
    maze=mh.parse_maze(maze_str)
    start=mh.find_pos(maze,'S')
    goal=mh.find_pos(maze,'G')
    free=int((maze!='X').sum())
    summary.append({'maze':maze_file,'rows':maze.shape[0],'columns':maze.shape[1],'free states':free,'start':start,'goal':goal})
    print(maze_file)
    print(f'  size={maze.shape}, free states={free}, start={start}, goal={goal}')
    mh.show_maze(maze)

display(pd.DataFrame(summary))
```
